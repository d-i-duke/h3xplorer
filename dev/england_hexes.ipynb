```python
import geopandas as gpd
import h3.api.numpy_int as h3
from lonboard import Map, PolygonLayer, ScatterplotLayer
```

```python
bottomleft = (49.970022, -5.918893)
bottomright = (51.213532, 1.632811)
topright = (58.721506, -2.206627)
topleft = (57.842317, -7.826258)
polygon = [topleft, topright, bottomright, bottomleft]
latlonpoly = h3.LatLngPoly(polygon)
```

```python
refs = h3.h3shape_to_cells(latlonpoly, 5)
```

```python
all_hexes = gpd.GeoDataFrame(
    {"h3_ref": refs, "geometry": [h3.cells_to_h3shape([ref]) for ref in refs]}, crs="EPSG:4326"
)
all_hexes
```

```python
points = gpd.GeoDataFrame(
    {"ref": [1, 2, 3, 4, 5]},
    geometry=gpd.points_from_xy(
        x=[-3.448079, -2.001428, -0.494362, -0.128329, -2.001375],
        y=[51.689821, 52.597808, 53.487243, 51.503992, 51.249166],
        crs="EPSG:4326",
    ),
)
```

```python
hex_layer = PolygonLayer.from_geopandas(all_hexes)
hex_layer.get_fill_color = [255, 0, 0, 100]
hex_layer.get_line_color = [255, 0, 0, 255]
hex_layer.line_width_min_pixels = 0.5
hex_layer.line_width_max_pixels = 5
point_layer = ScatterplotLayer.from_geopandas(points)
point_layer.get_radius = 500
m = Map([hex_layer, point_layer])
m
```
