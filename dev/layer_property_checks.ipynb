```python
import geopandas as gpd
import pandas as pd
from lonboard import Map

from h3xplorer import plotting
```

```python
gdf = gpd.GeoDataFrame(
    {"data": pd.Series([-5, 0, 5, 10])},
    geometry=gpd.points_from_xy([-0.5, 0, 0.5, 1], [51, 51.2, 50.8, 50.9]).buffer(
        0.1, resolution=1
    ),
    crs="EPSG:4326",
)
gdf
```

```python
for num in range(4):
    display(list(gdf.geometry[num].exterior.coords))
```

```python
layer = plotting.create_polygon_layer(gdf, "data")
layer
```

```python
[layer.get_fill_color[num].as_py() for num in range(4)]
```

```python
[layer.get_line_color[num].as_py() for num in range(4)]
```

```python
layer2 = plotting.create_polygon_layer(gdf, "data", max_threshold=20)
```

```python
[layer2.get_fill_color[num].as_py() for num in range(4)]
```

```python
Map([layer])
```

```python
Map([layer2])
```
